```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
from google.colab import drive
drive.mount('/content/drive')

file_path = '/content/drive/MyDrive/SupplyChainProject/Cleaned_DataCo.csv'
df = pd.read_csv(file_path)

print("Data loaded! Preparing features for Machine Learning...")

features = ['shipping_mode', 'market', 'order_item_quantity', 'product_price']
X = df[features]
y = df['is_late']
X_encoded = pd.get_dummies(X, columns=['shipping_mode', 'market'], drop_first=True)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

print("Training the Random Forest model... (this might take a minute or two)")
rf_model = RandomForestClassifier(n_estimators=50, random_state=42)
rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)

print("\n--- Model Performance ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))
```
